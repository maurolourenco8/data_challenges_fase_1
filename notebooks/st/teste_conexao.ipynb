```python
import sys
from pathlib import Path
import pandas as pd
import psycopg2

# Adiciona a raiz do projeto ao path
root_path = Path.cwd().resolve().parents[1]
if str(root_path) not in sys.path:
    sys.path.append(str(root_path))

# Importa as variáveis de conexão já carregadas pelo main.py
from main import db_host, db_name, db_password, db_port, db_schema, db_user

# Abre a conexão usando psycopg2
conn = psycopg2.connect(
    host=db_host,
    dbname=db_name,
    user=db_user,
    password=db_password,
    port=db_port,
    sslmode="require"  # Necessário para o Neon PostgreSQL
)

query = f"""
SELECT 
    table_schema AS schema_tabela,
    table_name AS nome_tabela,
    table_type AS tipo_tabela
FROM information_schema.tables
WHERE table_schema = '{db_schema}'
ORDER BY table_name;
"""

df_tabelas = pd.read_sql(query, conn)
conn.close()

df_tabelas
```
